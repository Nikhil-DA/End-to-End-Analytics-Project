# Step 7: Cleaning and Wrangling

```python
# merge with customers 
df = df_sales.merge(
    df_customers,
    how = 'left',
    left_on = 'Customer Name Index',
    right_on = 'Customer Index'
)
```

```python
df
```

```python
# merge with products
df = df.merge(
    df_products,
    how = 'left',
    left_on = 'Product Description Index',
    right_on = 'Index'
)
df.head()
```

```python
# merge with regions
df = df.merge(
    df_regions,
    how = 'left',
    left_on = 'Delivery Region Index',
    right_on = 'id'
)
df.head()
```

```python
# merge with State Regions
df = df.merge(
    df_state_reg[["State Code", "Region"]],
    how = 'left',
    left_on = 'state_code',
    right_on = 'State Code'
)
df.head()
```

```python
# merging with budgets

df = df.merge(
    df_budgets,
    how='left',
    on = 'Product Name'
)
```

```python
df.to_csv(r"D:\Data Analytics\End to End Data analytics Project\df_file.csv", index = False)
```

# Step 8: Clean Up Redundant Columns

After merging multiple tables (Sales, Customers, Products, Regions), we often end up with **duplicate or technical ID columns** that are no longer needed for analysis or visualization.

In this step, we'll drop the following columns:

- `Customer Index` → already merged, we can keep the name instead
- `Index` → from product table merge
- `id` → often an internal row ID
- `State Code` → used for joining, not needed in visuals

This helps **declutter the DataFrame** and makes it easier to work with.


```python
df.drop(columns=[
    'Customer Index',
    'Index',
    'id',
    'State Code'
], inplace=True)
df.head()
```

# Step 9: Standardize Column Headers to Snake Case

To make our column names easier to work with in Python, we convert them all to **snake_case** format:
- All lowercase
- Spaces and dashes replaced with underscores
- Leading/trailing spaces removed

This improves readability and reduces errors when referencing column names in code.


```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(' ', '_')
        .str.replace('-', '_')
    )
    return df
```

```python
df = clean_columns(df)
```

# Step 11: Blank Out Budgets for Non-2017 Orders

The `budget` column in our dataset is only relevant for the year **2017**.

To avoid misleading analysis, we will set the `budget` values to blank (missing) for all orders that do **not belong to 2017**.

This ensures that comparisons between actual revenue and budget are accurate and only apply to the correct time period.


```python
df.loc[df['order_date'].dt.year != 2017, 'budget'] = pd.NA

df[['order_date', 'product_name', 'revenue', 'budget']].head(5)
```
